```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
import numpy as np


try:
    data = pd.read_csv("data.csv")
    print("Dataset Loaded Sucessfully !!")
except FileNotFoundError:
    print("Dataset not found. Please check the file path.")
```

```python
print("=== Explore Dataset ===")
print("1. Display First 5 Rows")
print("2. Display Data Shape")
print("3. Display Data Types")
print("4. Display Data Infomation")

choice=int(input("Enter Your Choice : "))

try:    
    if choice==1:
        print("==== First 5 Rows ====")
        print(data.head())
    elif choice==2:
        print("==== Check Dataset Shape ====")
        print(data.shape)
    elif choice==3:
        print("==== Check Data-Types ====")
        print(data.dtypes)
    elif choice==4:
        print("==== Display Data-Infomation ====")
        data.info()
    else:
        print("Invalid Choice (1-4)")
except :
    print("Error Occured !! Please Check Your Code")
```

```python
print("=== Data Cleaning ===")
print("1. Handle Missing Values")
print("2. Remove Duplicate Records")
print("3. Convert Data Column into Date-Time Format")
print("4. Check Invalid Stock Price Values")


choice=int(input("Enter Your Choice : "))
try:
    if choice==1:
        print("==== Handle Missing Data ====")
        print("1. Get Sum Of Missing Values")
        print("2. Drop Rows with Missing Values")
        print("3. Replace Missing Values With Specific Value")

        ch=int(input("Enter Your Choice : "))

        if ch==1:
            print("===== Sum of Missing Values =====")
            print(data.isna().sum())
        elif ch==2:
            print("===== Drop Rows With Missing Values =====")
            if data.isnull().empty:
                print("No Missing Values in Dataset !!")
            else:
                print("====== Before Cleaning =====")
                print()
                print(data.isna().sum())

                print("====== After Cleaning ======")
                print()
                data=data.dropna()
                print(data.isna().sum())
                print()
                print("Data Cleaned Sucessfullly !!")
        elif ch==3:
            print("===== Replace Missing Values With Specific Value =====")
            print("1. Replace With 0")
            print("2. Replace With Mean")
            print("3. Replace With Median")
            print("4. Replace With Mode")

            ch1=int(input("Enter Your Choice : "))
            if ch1==1:
                data=data.fillna(0)
                print("Missing Values Replaced With 0 Successfully !!")
            elif ch1==2:
                data=data.fillna(data.mean())
                print("Missing Values Replaced With Mean Successfully !!")
            elif ch1==3:
                data=data.fillna(data.median())
                print("Missing Values Replaced With Median Successfully !!")
            elif ch1==4:
                data=data.fillna(data.mode().iloc[0])
                print("Missing Values Replaced With Mode Successfully !!")
            else:
                print("Invalid Choice (1-4)")
        

    elif choice==2:
        print("==== Remove Duplicate Records ====")
        print("1. Check For Duplicate Records")
        print("2. Remove Duplicate Records")

        ch=int(input("Enter Your Choice : "))

        if ch==1:
            print("===== Check For Duplicate Records =====")
            print(data.duplicated().sum())
        elif ch==2:
            print("===== Remove Duplicate Records =====")
            if data.duplicated().sum()==0:
                print("No Duplicate Records Found !!")
            else:
                print("====== Before Cleaning =====")
                print()
                print(data.duplicated().sum())

                print("====== After Cleaning ======")
                print()
                data=data.drop_duplicates()
                print(data.duplicated().sum())
                print()
                print("Duplicate Records Removed Sucessfully !!")
        else:
            print("Invalid Choice (1-2)")
    elif choice==3:
        print("==== Convert Data Column into Date-Time Format ====")
        data['Date'] = pd.to_datetime(data['Date'])
        print("Data Column Converted into Date-Time Format Successfully !!")
    elif choice==4:
        print("==== Check Invalid Stock Price Values ====")
        print("1. Check For Negative Stock Price Values")
        print("2. Check For Zero Stock Price Values")

        ch=int(input("Enter Your Choice : "))

        if ch==1:
            print("===== Check For Negative Stock Price Values =====")
            negative_values = data[(data['Open'] < 0) | (data['High'] < 0) | (data['Low'] < 0) | (data['Close'] < 0)]
            if negative_values.empty:
                print("No Negative Stock Price Values Found !!")
            else:
                print("Negative Stock Price Values Found !!")
                print(negative_values)
        elif ch==2:
            print("===== Check For Zero Stock Price Values =====")
            zero_values = data[(data['Open'] == 0) | (data['High'] == 0) | (data['Low'] == 0) | (data['Close'] == 0)]
            if zero_values.empty:
                print("No Zero Stock Price Values Found !!")
            else:
                print("Zero Stock Price Values Found !!")
                print(zero_values)
    else:
        print("Invalid Choice (1-4)")
except :
    print("Error Occured !! Please Check Your Code")
```

```python
print("==== Data Analysis ====")
print("1. Analyze Historical Stock Prices ")
print("2. Analyze Open Price ")
print("3. Analyze High Price ")
print("4. Analyze Low Price ")
print("5. Analyze Close Price ")
print("6. Analyze Adjusted Close Price ")
print("7. Analyze Trading Volume ")
print("8. Calculate Daily Stock Price Changes")
print("9. Calculate Moving Averages")
print("10. Identify Highest and Lowest Stock Prices")

choice=int(input("Enter Your Choice : "))

try:
    if choice==1:
        print("===== Analyze Historical Stock Prices =====")
        print(data[["date", "open", "high", "low", "close", "volume"]].describe())

    elif choice==2:
        print("===== Analyze Open Prices =====")
        print(data["Open"].agg(["min", "max", "mean"]))
    elif choice==3:
        print("===== Analyze High Prices =====")
        print(data["High"].agg(["min", "max", "mean"]))
    elif choice==4:
        print("===== Analyze Low Prices =====")
        print(data["Low"].agg(["min", "max", "mean"]))
    elif choice==5:
        print("===== Analyze Close Prices =====")
        print(data["Close"].agg(["min","max","mean"]))
    elif choice==6:
        print("===== Analyze Adjusted Close Prices =====")
        print(data['Adj_Close'].agg(["min","max","mean"]))
    elif choice==7:
        print("===== Analyze Trading Volume =====")
        print(data["Volume"].agg(["min","max","mean","sum"]))
    elif choice==8:
            print("===== Calculate Daily Stock Price Changes =====")
            data["daily_change"] = data["Close"] - data["Open"]

            data["daily_change_percent"] = (data["daily_change"] / data["Open"])*100

            print(data[['Date', 'Open', 'Close', 'daily_change', 'daily_change_percent']].head())
                
    elif choice==9:
        print("===== Calculate Moving Averages =====")
        data = data.sort_values("Date").copy()
        data["moving_average"] = data["Close"].rolling(7).mean()
        print(data[["Date", "Close", "moving_average"]].tail())

    elif choice==10:
        print("===== Identify Highest and Lowest Stock Prices =====")
        print("Highest price:", data["High"].max())
        print("Lowest price:", data["Low"].min())

    else:
        print("Invalid Choice (1-10)")
except :
    print("Error Occured !! Please Check Your Code")
```

```python
print("=== Data Visualizaation ===")
print("1. Line Chart : Closing Price Over Time")
print("2. Line Chart : Opening Price Over Time")
print("3. Line Chart : Moving Averages")
print("4. Bar Chart : Trading Volume")
print("5. Scatter Plot : Trading Volume Vs Closing Price ")

choice=int(input("Enter Your Choice : "))
try:
    if choice==1:
        print("===== Line Chart : Closing Price Over Time =====")
        plt.figure(figsize=(12,6))
        plt.plot(data["Date"], data["Close"], label="Closing Price", color="blue")
        plt.title("Closing Price Over Time")
        plt.xlabel("Date")
        plt.ylabel("Price")
        plt.legend()
        plt.grid()
        plt.show()
        print("Chart Displayed Successfully !!")

    elif choice==2:
        print("===== Line Chart : Opening Price Over Time =====")
        plt.figure(figsize=(12,6))
        plt.plot(data["Date"], data["Open"], label="Opening Price", color="green")
        plt.title("Opening Price Over Time")
        plt.xlabel("Date")
        plt.ylabel("Price")
        plt.legend()
        plt.grid()
        plt.show()
        print("Chart Displayed Successfully !!")

    elif choice==3:
        print("===== Line Chart : Moving Averages =====")
        data = data.sort_values("Date").copy()
        data["moving_average"] = data["Close"].rolling(7).mean()

        plt.figure(figsize=(12,6))
        plt.plot(data["Date"], data["Close"], label="Closing Price", color="blue")
        plt.plot(data["Date"], data["moving_average"], label="7-Day Moving Average", color="orange")
        plt.title("Closing Price and Moving Average Over Time")
        plt.xlabel("Date")
        plt.ylabel("Price")
        plt.legend()
        plt.grid()
except :
    print("Error Occured !! Please Check Your Code")
```

```python
print("=== Analysis Results ===")
print("1. Identify stock price trends.")
print("2. Compare stock performance over time.")
print("3. Analyze moving average patterns.")
print("4. Identify periods of high trading volume.")

choice=int(input("Enter Your Choice : "))
try:
    if choice == 1:
        print("\n========== STOCK PRICE TRENDS ==========")
        data["Moving_Average"] = data["Close"].rolling(7).mean()

        if data["Close"].iloc[-1] > data["Moving_Average"].iloc[-1]:
            print("Stock Trend: Upward")
        else:
            print("Stock Trend: Downward")
    elif choice == 2:
        print("\n========== STOCK PERFORMANCE ==========")
        first_price = data["Close"].iloc[0]
        last_price = data["Close"].iloc[-1]

        change = last_price - first_price
        percentage = (change / first_price) * 100

        print("Starting Price:", first_price)
        print("Latest Price:", last_price)
        print("Price Change:", change)
        print("Performance:", percentage, "%")

    elif choice == 3:
        print("\n========== MOVING AVERAGE PATTERNS ==========")
        data["Moving_Average"] = data["Close"].rolling(7).mean()

        plt.figure(figsize=(12, 6))
        plt.plot(data["Date"], data["Close"], label="Close Price")
        plt.plot(data["Date"], data["Moving_Average"], label="Moving Average")
        plt.title("Moving Average Patterns")
        plt.xlabel("Date")
        plt.ylabel("Price")
        plt.legend()
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()


    elif choice == 4:
        print("\n========== HIGH TRADING VOLUME PERIODS ==========")
        high_vol_threshold = data["Volume"].quantile(0.75)
        high_vol_periods = data[data["Volume"] > high_vol_threshold]

        if not high_vol_periods.empty:
            print("Periods of High Trading Volume:")
            print(high_vol_periods[["Date", "Volume"]])
        else:
            print("No periods of high trading volume found.")

    elif choice == 5:
        print("\n========== FINAL STOCK REPORT ==========")
        print(data[["Open", "High", "Low", "Close", "Volume"]].describe())
        print("\nOverall Performance:")

        if data["Close"].iloc[-1] > data["Close"].iloc[0]:
            print("Stock Price Increased During This Period.")
        elif data["Close"].iloc[-1] < data["Close"].iloc[0]:
            print("Stock Price Decreased During This Period.")
        else:
            print("Stock Price Remained Unchanged.")

    else:
        print("Invalid Choice (1-5)")
except :
    print("Error Occured !! Please Check Your Code")
```
